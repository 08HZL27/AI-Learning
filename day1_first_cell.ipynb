## 今天的知识转化：什么是向量？

盲写完之后，运行下面的 cell 看看 numpy 里的向量长什么样：

```python
import numpy as np

# 向量 = 一列有顺序的数字（知识库 [[101.1-向量]] 里的定义）
v = np.array([2, 5, 1])
w = np.array([3, -1, 4])

print("向量 v =", v)
print("向量 w =", w)
print("v + w =", v + w, "  ← 向量加法（对应坐标相加）")
print("点积 v·w =", v @ w, "  ← 点积衡量相似度（Attention 的核心）")
```
