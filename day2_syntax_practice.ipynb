## ① 循环（for / while / range）

```python
# 示例：for 循环 + range
for i in range(5):
    print("第", i, "次")

# 练习：打印 1 到 10 的平方（1, 4, 9, 16 ... 100）
# 提示：range(1, 11)
for i in range(1,11):
    print(i*i)
```

## ② 函数（def / 参数 / 返回值）

```python
# 示例：定义函数
def add(a, b):
    return a + b

print(add(3, 5))

# 练习：写一个函数 is_even(n)，判断 n 是偶数返回 True，奇数返回 False
# 然后打印 is_even(4) 和 is_even(7) 的结果

def is_even(n):
    return n % 2 == 0    # 直接返回判断结果（True/False）

print(is_even(7), is_even(4))   # 输出：False True


```

## ⑤ 文件读写（open / write / read）

```python
# 示例：写入 + 读取
with open("demo.txt", "w", encoding="utf-8") as f:
    f.write("第一行\n第二行\n")
with open("demo.txt", "r", encoding="utf-8") as f:
    print(f.read())

# 练习：把上面 3 个同学的成绩写入 score.txt（每行一个人），
# 再读出来打印（提示：open("score.txt", "w") 然后 read 回来）
score = {"张伟": 90, "李娜": 85, "王强": 78}

with open("score.txt", "w", encoding="utf-8") as f:
    for name, s in score.items():          # 循环遍历字典
        f.write(name + " " + str(s) + "\n")  # 每行写一个同学

with open("score.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

```python
import shutil
import os
count=0
folder = r"C:\Users\hzl\AI-Learning\01-python-scripts\test_folder"
for name in os.listdir(folder):
    full = os.path.join(folder, name)
    if os.path.isfile(full):
        ext = os.path.splitext(name)[1]            # 填空①：splitext 的 [1] 就是扩展名
        if ext:
            target = ext[1:]     # 填空②：切片去掉点号（第2个字符开始）
        else:
            target = "others"
        os.makedirs(os.path.join(folder, target), exist_ok=True)
        print(name, "→", target)
        count+=1
        shutil.move(full, os.path.join(folder, target, name))
print("整理完成，共移动", count, "个文件") 
```

```python
# ============================================
# 矩阵乘法三步走：从入门到"哦！"
# 规则只有一条：一行 × 一列，对号入座相乘，再全加起来
# ============================================
import numpy as np

# ---------- 第 1 关：成绩加权总分（最生活化） ----------
A = [
    [90, 80, 70],   # 学生1 的三科成绩
    [60, 70, 80],   # 学生2 的三科成绩
]
B = [
    [0.3],          # 平时 占 30%
    [0.2],          # 期中 占 20%
    [0.5],          # 期末 占 50%
]

print("========== 第 1 关：成绩加权 ==========")
print("学生1 总分 = 90×0.3 + 80×0.2 + 70×0.5")
print("           =", 90*0.3, "+", 80*0.2, "+", 70*0.5)
print("           =", 90*0.3 + 80*0.2 + 70*0.5)
print()
print("numpy 验证:", np.array(A) @ np.array(B))

# ---------- 第 2 关：你的小练习（学生3） ----------
print("\n========== 第 2 关：你的练习 ==========")
s3 = 80*0.3 + 90*0.2 + 60*0.5
print("学生3 总分 = 80×0.3 + 90×0.2 + 60×0.5 =", s3)

# ---------- 第 3 关：旋转+缩放（同一规则，换了数字） ----------
R = np.array([[0, -1],
              [1,  0]])     # 旋转 90°
S = np.array([[2, 0],
              [0, 2]])      # 放大 2 倍
p = np.array([1.0, 0.0])    # 测试点 (1, 0)

print("\n========== 第 3 关：旋转+缩放 ==========")
print("点 (1,0) 旋转后:", R @ p)
print("再放大 2 倍  :", S @ (R @ p))
print("合成一步     :", (S @ R) @ p)
print("\n结论：规则从头到尾就一条——一行×一列，相乘再相加")
```
